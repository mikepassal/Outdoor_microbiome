```python
from itertools import combinations

# Create a dictionary mapping GO terms to their genes
go_to_genes = {}
for go_term in go_terms:
    if go_term not in arabi_go_matrix.columns:
        print(f"Warning: GO term {go_term} not found in the matrix, skipping.")
        go_terms = go_terms[
            go_terms != go_term
        ]  # Remove the missing GO term from the list
        continue
    # Get genes that have this GO term (columns with value 1)
    genes = arabi_go_matrix[arabi_go_matrix[go_term] == 1].index.tolist()
    go_to_genes[go_term] = set(genes)
    print(f"{go_term}: {len(genes)} genes")

print(f"\nTotal GO terms: {len(go_to_genes)}")

# Generate all possible combinations of GO terms
# Get all combinations of different sizes
all_combinations = {}
for r in range(1, 9):
    print(len(list(combinations(go_terms, r))))
    for combo in combinations(go_terms, r):
        # Find union of genes for this combination
        gene_sets = [go_to_genes[term] for term in combo]
        combined_genes = set.union(*gene_sets)
        combo_key = " OR ".join(combo)
        all_combinations[combo_key] = combined_genes

# Filter to only keep combinations with more than 20 genes
filtered_combinations = {k: v for k, v in all_combinations.items() if len(v) > 5}

print(f"\nGenerated {len(all_combinations)} combinations")
print(f"Filtered to {len(filtered_combinations)} combinations with >20 genes")
print("\nCombinations with gene counts:")
```

```python
import numpy as np
from scipy.stats import pearsonr, spearmanr, rankdata

# Get Pseudomonas_E_647626 abundance for each sample (using 'Sample' which has LIC IDs)
pseudomonas_data = genus_sum_w_tp[genus_sum_w_tp["Genus"] == "Pseudomonas_E_647626"]
pseudomonas_abundance = pseudomonas_data.set_index("Sample")["AbundR100"]

# Get the mapping from filename to sampID (LIC IDs) from metadata
# We need samples that have both transcriptome data and Pseudomonas abundance
common_lic_ids = list(
    set(long_term_metadata["sampID"]) & set(pseudomonas_abundance.index)
)
common_filenames = long_term_metadata[
    long_term_metadata["sampID"].isin(common_lic_ids)
].index.tolist()

# Also filter to only those that are in the transcriptome
common_filenames = [f for f in common_filenames if f in long_term_transcriptome.index]

print(
    f"Found {len(common_filenames)} samples in common between transcriptome and microbiome data"
)

# Get Pseudomonas abundance vector (aligned to common_filenames)
lic_ids = long_term_metadata.loc[common_filenames, "sampID"]
pseudo_abund = pseudomonas_abundance.loc[lic_ids].values

# --- Vectorized correlation calculation ---
# Build a matrix of median expressions: rows = samples, cols = combinations
transcriptome_cols = set(long_term_transcriptome.columns)
combo_names = []
combo_gene_counts_total = []
combo_gene_counts_in_data = []
median_matrix_cols = []

for combo_name, gene_set in filtered_combinations.items():
    genes_in_data = list(set(gene_set) & transcriptome_cols)
    if len(genes_in_data) == 0:
        continue
    combo_names.append(combo_name)
    combo_gene_counts_total.append(len(gene_set))
    combo_gene_counts_in_data.append(len(genes_in_data))
    median_expr = (
        long_term_transcriptome.loc[common_filenames, genes_in_data]
        .median(axis=1)
        .values
    )
    median_matrix_cols.append(median_expr)

# median_matrix: shape (n_samples, n_combinations)
median_matrix = np.column_stack(median_matrix_cols)
n_samples, n_combos = median_matrix.shape

# Vectorized Pearson correlation using matrix operations
# Center the columns and pseudo_abund
mm_centered = median_matrix - median_matrix.mean(axis=0, keepdims=True)
pa_centered = pseudo_abund - pseudo_abund.mean()

# Pearson r = dot(x_centered, y_centered) / (||x_centered|| * ||y_centered||)
numerator = mm_centered.T @ pa_centered  # (n_combos,)
denom_x = np.sqrt((mm_centered**2).sum(axis=0))  # (n_combos,)
denom_y = np.sqrt((pa_centered**2).sum())
pearson_rs = numerator / (denom_x * denom_y)

# Vectorized Spearman correlation (rank-based Pearson)
ranked_matrix = np.apply_along_axis(rankdata, 0, median_matrix)
ranked_pa = rankdata(pseudo_abund)
rm_centered = ranked_matrix - ranked_matrix.mean(axis=0, keepdims=True)
rpa_centered = ranked_pa - ranked_pa.mean()
num_s = rm_centered.T @ rpa_centered
denom_sx = np.sqrt((rm_centered**2).sum(axis=0))
denom_sy = np.sqrt((rpa_centered**2).sum())
spearman_rs = num_s / (denom_sx * denom_sy)

# Compute p-values from correlation coefficients using t-distribution
from scipy.stats import t as t_dist

df = n_samples - 2
# Pearson p-values
t_pearson = pearson_rs * np.sqrt(df / (1 - pearson_rs**2 + 1e-300))
pearson_pvals = 2 * t_dist.sf(np.abs(t_pearson), df)
# Spearman p-values
t_spearman = spearman_rs * np.sqrt(df / (1 - spearman_rs**2 + 1e-300))
spearman_pvals = 2 * t_dist.sf(np.abs(t_spearman), df)

# Build results DataFrame
correlation_df = pd.DataFrame(
    {
        "combination": combo_names,
        "n_genes_total": combo_gene_counts_total,
        "n_genes_in_data": combo_gene_counts_in_data,
        "pearson_r": pearson_rs,
        "pearson_pval": pearson_pvals,
        "spearman_r": spearman_rs,
        "spearman_pval": spearman_pvals,
    }
)
correlation_df["abs_pearson"] = correlation_df["pearson_r"].abs()
correlation_df = correlation_df.sort_values("abs_pearson", ascending=False)

print(f"\nAnalyzed {len(correlation_df)} gene set combinations")
print("\nTop 20 combinations by absolute Pearson correlation:")
print(correlation_df.head(20).to_string())
```

# Individual Gene Correlations with Pseudomonas
Instead of summarizing gene sets by their median/mean expression, calculate each gene's independent correlation with Pseudomonas abundance.

```python
# Collect all unique genes directly from the defense term gene sets
all_genes_of_interest = set().union(*go_to_genes.values())
gene_to_terms = {}  # Track which GO terms each gene belongs to
for go_term, genes in go_to_genes.items():
    for gene in genes:
        gene_to_terms.setdefault(gene, []).append(go_term)

# Filter to genes present in the transcriptome
genes_in_transcriptome = sorted(
    list(all_genes_of_interest & set(long_term_transcriptome.columns))
)
print(f"Total unique genes across all defense terms: {len(all_genes_of_interest)}")
print(f"Genes present in transcriptome: {len(genes_in_transcriptome)}")

# --- Vectorized individual gene correlations with Pseudomonas_E_647626 ---
# Expression matrix: shape (n_samples, n_genes)
expr_matrix = long_term_transcriptome.loc[
    common_filenames, genes_in_transcriptome
].values
n_samp, n_genes = expr_matrix.shape

# Pearson correlation (vectorized across all genes at once)
expr_centered = expr_matrix - expr_matrix.mean(axis=0, keepdims=True)
pa_c = pseudo_abund - pseudo_abund.mean()
numer = expr_centered.T @ pa_c
denom_expr = np.sqrt((expr_centered**2).sum(axis=0))
denom_pa = np.sqrt((pa_c**2).sum())
gene_pearson_rs = numer / (denom_expr * denom_pa)

# Spearman correlation (vectorized)
ranked_expr = np.apply_along_axis(rankdata, 0, expr_matrix)
ranked_pa_vec = rankdata(pseudo_abund)
re_centered = ranked_expr - ranked_expr.mean(axis=0, keepdims=True)
rpa_c = ranked_pa_vec - ranked_pa_vec.mean()
num_sp = re_centered.T @ rpa_c
denom_re = np.sqrt((re_centered**2).sum(axis=0))
denom_rpa = np.sqrt((rpa_c**2).sum())
gene_spearman_rs = num_sp / (denom_re * denom_rpa)

# P-values
df_gene = n_samp - 2
t_pear = gene_pearson_rs * np.sqrt(df_gene / (1 - gene_pearson_rs**2 + 1e-300))
gene_pearson_pvals = 2 * t_dist.sf(np.abs(t_pear), df_gene)
t_spear = gene_spearman_rs * np.sqrt(df_gene / (1 - gene_spearman_rs**2 + 1e-300))
gene_spearman_pvals = 2 * t_dist.sf(np.abs(t_spear), df_gene)

# Build results DataFrame
individual_gene_corr_df = pd.DataFrame(
    {
        "gene": genes_in_transcriptome,
        "pearson_r": gene_pearson_rs,
        "pearson_pval": gene_pearson_pvals,
        "spearman_r": gene_spearman_rs,
        "spearman_pval": gene_spearman_pvals,
    }
)
individual_gene_corr_df["abs_pearson"] = individual_gene_corr_df["pearson_r"].abs()
individual_gene_corr_df["member_of_go_terms"] = [
    "; ".join(gene_to_terms.get(g, [])) for g in genes_in_transcriptome
]
individual_gene_corr_df = individual_gene_corr_df.sort_values(
    "abs_pearson", ascending=False
)

print(f"\nIndividual gene correlations with Pseudomonas_E_647626:")
print(f"Analyzed {len(individual_gene_corr_df)} genes")
print("\nTop 30 genes by absolute Pearson correlation:")
print(individual_gene_corr_df.head(30).to_string())
```

## Lineplots of immunity gene sets across long-term samples
These plots show per-gene expression trajectories across all samples in `long_term_transcriptome` for each predefined gene set.

```python
import matplotlib.pyplot as plt


def plot_gene_set_lines(expr_df, gene_set, title):
    genes_present = [g for g in gene_set if g in expr_df.columns]
    missing_genes = [g for g in gene_set if g not in expr_df.columns]

    if not genes_present:
        print(f"No genes from '{title}' are present in long_term_transcriptome.")
        return

    plt.figure(figsize=(14, 6))
    for gene in genes_present:
        gene_values = expr_df[gene].values.astype(float)
        gene_min = gene_values.min()
        gene_max = gene_values.max()
        if gene_max > gene_min:
            gene_values = (gene_values - gene_min) / (gene_max - gene_min)
        else:
            gene_values = gene_values * 0
        plt.plot(expr_df.index, gene_values, label=gene, linewidth=1.5, alpha=0.9)

    plt.title(title)
    plt.xlabel("Sample")
    plt.ylabel("Normalized expression (min-max, 0-1)")
    plt.tick_params(axis="x", which="both", bottom=False, labelbottom=False)
    plt.legend(title="Gene", bbox_to_anchor=(1.02, 1), loc="upper left")
    plt.tight_layout()
    plt.show()

    if missing_genes:
        print(f"Missing genes in transcriptome for '{title}': {missing_genes}")


plot_gene_set_lines(
    long_term_transcriptome,
    priming_only_genes,
    "Priming Only Genes Across Long-Term Samples",
)
plot_gene_set_lines(
    long_term_transcriptome,
    priming_and_post_challenge_genes,
    "Priming and Post-Challenge Genes Across Long-Term Samples",
)
plot_gene_set_lines(
    long_term_transcriptome,
    rechallenge_genes,
    "Rechallenge Genes Across Long-Term Samples",
)
```

```python
from itertools import combinations
from scipy.stats import spearmanr


def compute_pairwise_spearman(expr_df, gene_set):
    genes_present = [g for g in gene_set if g in expr_df.columns]
    missing_genes = [g for g in gene_set if g not in expr_df.columns]

    pair_rows = []
    for gene_a, gene_b in combinations(genes_present, 2):
        rho, pval = spearmanr(expr_df[gene_a], expr_df[gene_b], nan_policy="omit")
        pair_rows.append(
            {
                "gene_a": gene_a,
                "gene_b": gene_b,
                "spearman_rho": rho,
                "spearman_pval": pval,
                "abs_rho": abs(rho),
            }
        )

    pairwise_df = pd.DataFrame(pair_rows).sort_values("abs_rho", ascending=False)
    corr_matrix = expr_df[genes_present].corr(method="spearman")

    return genes_present, missing_genes, pairwise_df, corr_matrix


set_map = {
    "priming_only_genes": priming_only_genes,
    "priming_and_post_challenge_genes": priming_and_post_challenge_genes,
    "rechallenge_genes": rechallenge_genes,
}

pairwise_spearman_results = {}
for set_name, gene_set in set_map.items():
    genes_present, missing_genes, pairwise_df, corr_matrix = compute_pairwise_spearman(
        long_term_transcriptome, gene_set
    )

    pairwise_spearman_results[set_name] = {
        "genes_present": genes_present,
        "missing_genes": missing_genes,
        "pairwise_table": pairwise_df,
        "correlation_matrix": corr_matrix,
    }

    print(f"\n=== {set_name} ===")
    print(f"Genes present: {len(genes_present)}")
    if missing_genes:
        print(f"Missing genes: {missing_genes}")

    print(f"Pairwise comparisons: {len(pairwise_df)}")
    if len(pairwise_df) > 0:
        print("Top correlations by |rho|:")
        display(
            pairwise_df[["gene_a", "gene_b", "spearman_rho", "spearman_pval"]].head(10)
        )
        print("Spearman correlation matrix:")
        display(corr_matrix)

# Example access:
# pairwise_spearman_results["priming_only_genes"]["pairwise_table"]
# pairwise_spearman_results["priming_only_genes"]["correlation_matrix"]
```

```python
from scipy.stats import pearsonr, spearmanr
import numpy as np

# Ensure FLG Arabidopsis genes are available
flg_arabidopsis_genes = (
    flg_genes["Arabidopsis_thaliana"].dropna().astype(str).str.strip()
)
flg_arabidopsis_genes = [
    gene for gene in flg_arabidopsis_genes if gene not in ["", "nan", "None"]
]
flg_arabidopsis_genes = list(dict.fromkeys(flg_arabidopsis_genes))

# Targets aligned to transcriptome filename index
ham16s_by_filename = long_term_metadata_w_load_info["ham16s"].copy()

lic_to_filename = pd.Series(
    long_term_metadata.index, index=long_term_metadata["sampID"]
)
pseudomonas_e_by_filename = pseudomonas_abundance.copy()
pseudomonas_e_by_filename.index = lic_to_filename.reindex(
    pseudomonas_e_by_filename.index
).values
pseudomonas_e_by_filename = pseudomonas_e_by_filename.dropna()

flg_genes_present = [
    g for g in flg_arabidopsis_genes if g in long_term_transcriptome.columns
]
flg_genes_missing = [
    g for g in flg_arabidopsis_genes if g not in long_term_transcriptome.columns
]


def safe_corr(x, y):
    if len(x) < 3 or len(y) < 3:
        return np.nan, np.nan, np.nan, np.nan
    if pd.Series(x).nunique(dropna=True) < 2 or pd.Series(y).nunique(dropna=True) < 2:
        return np.nan, np.nan, np.nan, np.nan

    pearson_r, pearson_p = pearsonr(x, y)
    spearman_rho, spearman_p = spearmanr(x, y)
    return pearson_r, pearson_p, spearman_rho, spearman_p


rows = []
for gene in flg_genes_present:
    gene_series = long_term_transcriptome[gene]

    idx_load = gene_series.index.intersection(ham16s_by_filename.index)
    df_load = pd.DataFrame(
        {
            "gene_expr": gene_series.loc[idx_load],
            "target": ham16s_by_filename.loc[idx_load],
        }
    ).dropna()

    pearson_load_r, pearson_load_p, spearman_load_rho, spearman_load_p = safe_corr(
        df_load["gene_expr"], df_load["target"]
    )

    idx_pseudo = gene_series.index.intersection(pseudomonas_e_by_filename.index)
    df_pseudo = pd.DataFrame(
        {
            "gene_expr": gene_series.loc[idx_pseudo],
            "target": pseudomonas_e_by_filename.loc[idx_pseudo],
        }
    ).dropna()

    pearson_pseudo_r, pearson_pseudo_p, spearman_pseudo_rho, spearman_pseudo_p = (
        safe_corr(df_pseudo["gene_expr"], df_pseudo["target"])
    )

    rows.append(
        {
            "gene": gene,
            "n_samples_load": len(df_load),
            "pearson_r_load": pearson_load_r,
            "pearson_pval_load": pearson_load_p,
            "spearman_rho_load": spearman_load_rho,
            "spearman_pval_load": spearman_load_p,
            "n_samples_pseudomonas": len(df_pseudo),
            "pearson_r_pseudomonas": pearson_pseudo_r,
            "pearson_pval_pseudomonas": pearson_pseudo_p,
            "spearman_rho_pseudomonas": spearman_pseudo_rho,
            "spearman_pval_pseudomonas": spearman_pseudo_p,
        }
    )

flg_gene_correlations_df = pd.DataFrame(rows)
flg_gene_correlations_df = flg_gene_correlations_df.sort_values(
    "spearman_rho_pseudomonas", ascending=False
)

print(f"Total FLG Arabidopsis genes listed: {len(flg_arabidopsis_genes)}")
print(f"FLG genes found in transcriptome: {len(flg_genes_present)}")
print(f"FLG genes missing from transcriptome: {len(flg_genes_missing)}")
print("Created dataframe: flg_gene_correlations_df")
display(flg_gene_correlations_df.head(20))

# Optional save:
# flg_gene_correlations_df.to_csv(
#     "/Users/michael/Data/Luke_terrace_experiment/General_data/Junk_data/flg_gene_correlations_load_and_pseudomonas.csv",
#     index=False,
# )
```
